Voglio calcolare quanto ha perso in una seduta un portafoglio con il 60% in SPY e il 40% in IEF.

```python
# Giorno 1 — Dai prezzi alla perdita di portafoglio

# Calcolo il rendimento e la perdita giornaliera di un portafoglio
# con il 60% in SPY e il 40% in IEF.

spy_start = 100
spy_end = 98
ief_start = 100
ief_end = 101

spy_return = spy_end / spy_start - 1
ief_return = ief_end / ief_start - 1

print(f"Rendimento SPY: {spy_return:.2%}")
print(f"Rendimento IEF: {ief_return:.2%}")

spy_weight = 0.60
ief_weight = 0.40

spy_contribution = spy_weight * spy_return
ief_contribution = ief_weight * ief_return

portfolio_return = spy_contribution + ief_contribution

print(f"Rendimento P2: {portfolio_return:.2%}")
```

```python
portfolio_loss = -portfolio_return
starting_value = 100_000
loss_usd = starting_value * portfolio_loss

print(f"Perdita P2: {portfolio_loss:.2%}")
print(f"Perdita in dollari: {loss_usd:.2f} USD")
```

```python
ief_end_variant = 102
ief_return_variant = ief_end_variant / ief_start - 1

ief_contribution_variant = ief_weight * ief_return_variant
portfolio_return_variant = spy_contribution + ief_contribution_variant

portfolio_loss_variant = -portfolio_return_variant
loss_usd_variant = starting_value * portfolio_loss_variant

print(f"Rendimento IEF: {ief_return_variant:.2%}")
print(f"Rendimento P2: {portfolio_return_variant:.2%}")
print(f"Perdita P2: {portfolio_loss_variant:.2%}")
print(f"Perdita in dollari: {loss_usd_variant:.2f} USD")
```

```python
print(f"IEF return: {ief_return_variant:.2%}")
print(f"P2 return: {portfolio_return_variant:.2%}")
print(f"P2 loss: {portfolio_loss_variant:.2%}")
print(f"P2 loss in USD: {loss_usd_variant:.2f}")
```

### Why the portfolio lost $400

SPY fell by 2%, while IEF rose by 2%. The portfolio started the day with
$60,000 invested in SPY and $40,000 invested in IEF. The SPY position lost
$1,200, and the IEF position gained $800. The portfolio therefore ended
the day with a net loss of $400, equivalent to 0.40% of its starting value.
